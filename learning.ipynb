```python
import yfinance as yf
import matplotlib.pyplot as plt

df = yf.download('AAPL', start='2024-01-01', end='2024-12-31')

df['MA20'] = df['Close'].rolling(window=20).mean()
df['MA50'] = df['Close'].rolling(window=50).mean()

# Plot both
df['Close'].plot(label='Close', figsize=(12,5))
df['MA20'].plot(label='20-day MA')
df['MA50'].plot(label='50-day MA')

plt.legend()
plt.title('AAPL 2024 - Close vs Moving Average')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Create a signal column: 1 when MA20 > MA50, 0 otherwise
df['Signal'] = 0
df.loc[df['MA20'] > df['MA50'], 'Signal'] = 1

# Detect crossovers - where the signal CHANGES
df['Crossover'] = df['Signal'].diff()

# Plot
fig, ax = plt.subplots(figsize=(12,5))
ax.plot(df.index, df['Close'], label='Close', alpha=0.5)
ax.plot(df.index, df['MA20'], label='MA20')
ax.plot(df.index, df['MA50'], label='MA50')

# Golden Cross (buy) - green triangles pointing up
ax.plot(df[df['Crossover'] == 1].index, df['MA20'][df['Crossover'] == 1],
        '^', color='green', markersize=12, label='Golden Cross (Buy)')

# Death Cross (sell) - red triangles pointing down
ax.plot(df[df['Crossover'] == -1].index, df['MA20'][df['Crossover'] == -1],
        'v', color='red', markersize=12, label='Death Cross (Sell)')

ax.legend()
ax.set_title('AAPL 2024 - Golden & Death Cross Signals')
plt.show()
```

```python
import yfinance as yf
import matplotlib.pyplot as plt

df = yf.download('AAPL', start='2024-01-01', end='2024-12-31')

# Calculate daily price change
delta = df['Close'].diff()

# Separate gains and losses
gain = delta.clip(lower=0)
loss = -delta.clip(upper=0)

# 14-day average gain and loss
avg_gain = gain.rolling(window=14).mean()
avg_loss = loss.rolling(window=14).mean()

# RSI formula
rs = avg_gain / avg_loss
df['RSI'] = 100 - (100 / (1+ rs))

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

ax1.plot(df.index, df['Close'], label='Close')
ax1.set_title('AAPL 2024 - Close Price')
ax1.legend()

ax2.plot(df.index, df['RSI'], label='RSI', color='purple')
ax2.axhline(70, color='red', linestyle='--', label='Overbought (70)')
ax2.axhline(30, color='green', linestyle='--', label='Oversold (30)')
ax2.set_title('RSI')
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
import yfinance as yf
import matplotlib.pyplot as plt
import pandas as pd

# years = [('2022-01-01', '2022-12-31'),
#          ('2023-01-01', '2023-12-31'),
#          ('2024-01-01', '2024-12-31')]
tickers = ['AAPL', 'MSFT', 'GOOGL', 'SPY', 'TSLA']

RSI_THRESHOLD = 45
STOP_LOSS = 0.06

results = []

for ticker in tickers:
    df = yf.download(ticker, start='2024-01-01', end='2024-12-31')
    df.columns = df.columns.droplevel(1)

    # Calculate MA50
    df['MA50'] = df['Close'].rolling(window=50).mean()

    # Calculate RSI
    delta = df['Close'].diff()
    gain = delta.clip(lower=0)
    loss = delta.clip(upper=0).abs()
    avg_gain = gain.rolling(window=14).mean()
    avg_loss = loss.rolling(window=14).mean()
    rs = avg_gain / avg_loss
    df['RSI'] = 100 - (100 / (1 + rs))

    # Calculate Buy/Sell signals and triggers
    # Combined signal: RSI oversold AND price above MA50
    df['Buy_Signal'] = (
        (df['RSI'] < RSI_THRESHOLD) &
        (df['Close'] > df['MA50'])
    ).astype(int)

    # Sell Signal: RSI Overbought
    df['Sell_Signal'] = (df['RSI'] > 70).astype(int)

    # Detect crossovers
    df['Buy_Trigger'] = df['Buy_Signal'].diff()
    df['Sell_Trigger'] = df['Sell_Signal'].diff()

    # Run trade loop
    trades = []
    in_trade = False
    entry_price = None
    entry_date = None
    for date, row in df.iterrows():
        if not in_trade:
            if row['Buy_Trigger'] == 1:
                in_trade = True
                entry_price = float(row['Close'])
                entry_date = date
        else:
            # Check stop-loss first
            if float(row['Close']) <= entry_price * (1 - STOP_LOSS):
                trades.append({
                    'entry_date': entry_date,
                    'exit_date': date,
                    'entry_price': entry_price,
                    'exit_price': row['Close'],
                    'exit_reason': 'stop-loss'
                })
                in_trade = False
    
            # Otherwise check normal sell signal
            elif row['Sell_Trigger'] == 1:
                trades.append({
                    'entry_date': entry_date,
                    'exit_date': date,
                    'entry_price': entry_price,
                    'exit_price': row['Close'],
                    'exit_reason': 'signal'
                })
                in_trade = False

    # Calculate total return and num_trades
    trades_df = pd.DataFrame(trades)
    if not trades_df.empty:
        trades_df['return_pct'] = (trades_df['exit_price'] - trades_df['entry_price']) / trades_df['entry_price'] * 100
        total_return = trades_df['return_pct'].sum()
        num_trades = len(trades_df)
    else:
        total_return = 0
        num_trades = 0

    # results.append({
    #     'year': start[:4],
    #     'total_return': total_return,
    #     'num_trades': num_trades,
    #     'stop_loss_exits': (trades_df['exit_reason'] == 'stop-loss').sum() if not trades_df.empty else 0,
    #     'signal_exits': (trades_df['exit_reason'] == 'signal').sum() if not trades_df.empty else 0
    # })
    results.append({'ticker': ticker, 
                    'total_return': total_return,
                    'num_trades': num_trades,
                    'stop_loss_exits': (trades_df['exit_reason'] == 'stop-loss').sum() if not trades_df.empty else 0,
                    'signal_exits': (trades_df['exit_reason'] == 'signal').sum() if not trades_df.empty else 0
                   })

results_df = pd.DataFrame(results)
print(results_df)
```
